```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))

```

E. Refleksi

**1. Perbedaan Supervised, Unsupervised, dan Reinforcement Learning**

* **Supervised Learning**
  * Menggunakan data yang sudah memiliki label (*input* dan target jawaban).
  * Bertujuan memprediksi label atau nilai pada data baru.
  * Contoh: Prediksi harga rumah dan deteksi email *spam*.
* **Unsupervised Learning**
  * Menggunakan data tanpa label (*input* saja).
  * Bertujuan mencari pola, struktur, atau pengelompokan (*cluster*) tersembunyi dari data.
  * Contoh: Segmentasi pelanggan berdasarkan perilaku belanja.
* **Reinforcement Learning**
  * Menggunakan sistem *agent* yang belajar melalui interaksi dengan lingkungan (*environment*).
  * Bertujuan mengambil keputusan terbaik berdasarkan umpan balik berupa *reward* (hadiah) dan *penalty* (hukuman).
  * Contoh: AI pemutar catur dan navigasi mobil otonom.

**2. Alasan Pembagian Data Latih dan Uji Itu Penting**

* **Menilai Kemampuan Generalisasi:** Memastikan model tidak hanya menghafal data latihan, tetapi juga bisa membuat prediksi yang akurat pada data baru yang belum pernah dilihat (*unseen data*).
* **Mencegah Overfitting:** Menghindari kondisi di mana model sangat baik pada data latih, tetapi performanya anjlok saat diuji dengan data baru.
* **Simulasi Performa Dunia Nyata:** Data uji berfungsi sebagai tolok ukur objektif untuk menguji keandalan model sebelum diterapkan secara langsung.

**3. Alasan Alur Kerja Machine Learning (ML) Disebut Iteratif**

* **Perlu Eksperimen Berulang:** Hasil prediksi model pada percobaan pertama jarang langsung optimal, sehingga butuh evaluasi dan penyesuaian berkali-kali.
* **Proses Perbaikan Terus-menerus:** Pengembang sering kali harus kembali ke tahap awal untuk membersihkan data, merkayasa fitur (feature engineering), memilih algoritma lain, atau mengatur ulang parameter (hyperparameter tuning).
* **Adaptasi Terhadap Perubahan Data:** Model yang sudah jadi perlu dilatih kembali secara berkala agar bisa menyesuaikan dengan pola data baru yang terus berkembang dari waktu ke waktu.
